```python
import os
import json
import yaml
import joblib
import random
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# STEP 2: DEFINE PATHS
# ============================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Manufacturing Company Capital and Registration Trend Analysis"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "4dbe5667-7b6b-41d7-82af-211562424d9a.csv"
)

os.makedirs(PROJECT_DIR, exist_ok=True)

H5_PATH = os.path.join(
    PROJECT_DIR, "manufacturing_model.h5"
)

PKL_PATH = os.path.join(
    PROJECT_DIR, "manufacturing_model.pkl"
)

YAML_PATH = os.path.join(
    PROJECT_DIR, "manufacturing_config.yaml"
)

JSON_PATH = os.path.join(
    PROJECT_DIR, "manufacturing_metadata.json"
)

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)

# ============================================================
# STEP 3: LOAD DATASET
# ============================================================

if not os.path.isfile(CSV_PATH):
    raise FileNotFoundError(
        f"CSV file not found: {CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH)

print("=" * 60)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 60)

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst five rows:")
print(df.head())

print("\nMissing values:")
print(df.isnull().sum())

# ============================================================
# STEP 4: DATA PREPROCESSING
# ============================================================

df.columns = df.columns.str.strip()

required_columns = [
    "AuthorizedCapital",
    "PaidupCapital",
    "CompanyRegistrationdate_date"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns missing: {missing_columns}"
    )

# Convert financial values to numeric

df["AuthorizedCapital"] = pd.to_numeric(
    df["AuthorizedCapital"],
    errors="coerce"
)

df["PaidupCapital"] = pd.to_numeric(
    df["PaidupCapital"],
    errors="coerce"
)

# Convert registration date

df["CompanyRegistrationdate_date"] = pd.to_datetime(
    df["CompanyRegistrationdate_date"],
    errors="coerce"
)

# Remove rows without target values

df = df.dropna(subset=["PaidupCapital"]).copy()

if len(df) < 4:
    raise ValueError(
        "At least four valid records are required."
    )

# ============================================================
# STEP 5: FEATURE ENGINEERING
# ============================================================

df["RegistrationYear"] = (
    df["CompanyRegistrationdate_date"].dt.year
)

df["RegistrationMonth"] = (
    df["CompanyRegistrationdate_date"].dt.month
)

df["RegistrationDay"] = (
    df["CompanyRegistrationdate_date"].dt.day
)

# Capital utilization is for descriptive analysis only.
# Do not use it as a predictor because it contains the target.

df["CapitalUtilization"] = np.where(
    df["AuthorizedCapital"] > 0,
    (
        df["PaidupCapital"] /
        df["AuthorizedCapital"]
    ) * 100,
    np.nan
)

numeric_features = [
    "AuthorizedCapital",
    "RegistrationYear",
    "RegistrationMonth",
    "RegistrationDay"
]

categorical_candidates = [
    "CompanyROCcode",
    "CompanyCategory",
    "CompanySubCategory",
    "CompanyClass",
    "Listingstatus",
    "CompanyStatus",
    "CompanyStateCode",
    "CompanyIndian/Foreign Company",
    "CompanyIndustrialClassification"
]

categorical_features = [
    col for col in categorical_candidates
    if col in df.columns
]

feature_columns = (
    numeric_features + categorical_features
)

X = df[feature_columns].copy()
y = df["PaidupCapital"].astype(float)

# Convert categories to strings while retaining missing data

for col in categorical_features:
    X[col] = X[col].astype("string").fillna("Missing")
    X[col] = X[col].astype(str)

print("\nSelected input features:")
print(feature_columns)

print("\nTarget variable: PaidupCapital")

# ============================================================
# STEP 6: CREATE PREPROCESSING PIPELINE
# ============================================================

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

# Compatible with recent and older scikit-learn versions

try:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            encoder
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            numeric_transformer,
            numeric_features
        ),
        (
            "categorical",
            categorical_transformer,
            categorical_features
        )
    ],
    remainder="drop"
)

# ============================================================
# STEP 7: TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

# ============================================================
# STEP 8: TRAIN RANDOM FOREST MODEL
# ============================================================

rf_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=100,
                max_depth=4,
                random_state=RANDOM_STATE
            )
        )
    ]
)

rf_pipeline.fit(X_train, y_train)

rf_predictions = rf_pipeline.predict(X_test)

rf_mae = mean_absolute_error(
    y_test, rf_predictions
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test, rf_predictions
    )
)

rf_r2 = (
    r2_score(y_test, rf_predictions)
    if len(y_test) >= 2
    else None
)

print("\n" + "=" * 60)
print("RANDOM FOREST RESULTS")
print("=" * 60)

print("MAE:", rf_mae)
print("RMSE:", rf_rmse)
print("R2 Score:", rf_r2)

# ============================================================
# STEP 9: CROSS-VALIDATION
# ============================================================

# This provides an exploratory evaluation.
# With 10 records, scores may vary substantially.

n_splits = min(5, len(df))

cv = KFold(
    n_splits=n_splits,
    shuffle=True,
    random_state=RANDOM_STATE
)

cv_mae_scores = -cross_val_score(
    rf_pipeline,
    X,
    y,
    cv=cv,
    scoring="neg_mean_absolute_error"
)

print("\nCross-validation MAE scores:")
print(cv_mae_scores)

print(
    "Mean CV MAE:",
    cv_mae_scores.mean()
)

# ============================================================
# STEP 10: PREPARE DATA FOR NEURAL NETWORK
# ============================================================

# Fit preprocessing on training data only

nn_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(strategy="median")
                ),
                (
                    "scaler",
                    StandardScaler()
                )
            ]),
            numeric_features
        ),
        (
            "categorical",
            Pipeline([
                (
                    "imputer",
                    SimpleImputer(
                        strategy="most_frequent"
                    )
                ),
                (
                    "encoder",
                    OneHotEncoder(
                        handle_unknown="ignore",
                        sparse_output=False
                    )
                )
            ]),
            categorical_features
        )
    ]
)

X_train_nn = nn_preprocessor.fit_transform(
    X_train
)

X_test_nn = nn_preprocessor.transform(
    X_test
)

X_train_nn = np.asarray(
    X_train_nn,
    dtype=np.float32
)

X_test_nn = np.asarray(
    X_test_nn,
    dtype=np.float32
)

# Scale target values for stable neural-network training

target_scaler = StandardScaler()

y_train_nn = target_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
).astype(np.float32)

# ============================================================
# STEP 11: BUILD NEURAL NETWORK
# ============================================================

input_dimension = X_train_nn.shape[1]

nn_model = Sequential([
    Input(shape=(input_dimension,)),
    Dense(32, activation="relu"),
    Dense(16, activation="relu"),
    Dense(8, activation="relu"),
    Dense(1, activation="linear")
])

nn_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="mse",
    metrics=["mae"]
)

print("\nNeural Network Architecture:")
nn_model.summary()

# ============================================================
# STEP 12: TRAIN NEURAL NETWORK
# ============================================================

early_stopping = EarlyStopping(
    monitor="loss",
    patience=30,
    restore_best_weights=True
)

history = nn_model.fit(
    X_train_nn,
    y_train_nn,
    epochs=200,
    batch_size=2,
    verbose=0,
    callbacks=[early_stopping]
)

print("\nNeural network training completed.")

# ============================================================
# STEP 13: EVALUATE NEURAL NETWORK
# ============================================================

nn_predictions_scaled = nn_model.predict(
    X_test_nn,
    verbose=0
)

nn_predictions = target_scaler.inverse_transform(
    nn_predictions_scaled
).flatten()

nn_mae = mean_absolute_error(
    y_test,
    nn_predictions
)

nn_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        nn_predictions
    )
)

nn_r2 = (
    r2_score(y_test, nn_predictions)
    if len(y_test) >= 2
    else None
)

print("\n" + "=" * 60)
print("NEURAL NETWORK RESULTS")
print("=" * 60)

print("MAE:", nn_mae)
print("RMSE:", nn_rmse)
print("R2 Score:", nn_r2)

# ============================================================
# STEP 14: SAVE H5 MODEL
# ============================================================

nn_model.save(H5_PATH)

print("\nH5 model saved:")
print(H5_PATH)

# ============================================================
# STEP 15: SAVE PKL FILE
# ============================================================

# Store both pipelines and neural-network preprocessing.
# Only load PKL files from trusted sources.

model_bundle = {
    "random_forest_pipeline": rf_pipeline,
    "neural_network_preprocessor": nn_preprocessor,
    "neural_network_target_scaler": target_scaler,
    "feature_columns": feature_columns,
    "numeric_features": numeric_features,
    "categorical_features": categorical_features,
    "target_column": "PaidupCapital"
}

joblib.dump(
    model_bundle,
    PKL_PATH
)

print("\nPKL file saved:")
print(PKL_PATH)

# ============================================================
# STEP 16: SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project_name": (
        "AI-Based Manufacturing Company Capital "
        "and Registration Trend Analysis"
    ),
    "dataset": {
        "file_path": CSV_PATH,
        "total_records": int(len(df)),
        "total_features": int(len(feature_columns))
    },
    "machine_learning": {
        "target": "PaidupCapital",
        "random_state": RANDOM_STATE,
        "test_size": 0.20,
        "models": [
            "RandomForestRegressor",
            "NeuralNetwork"
        ]
    },
    "random_forest": {
        "n_estimators": 100,
        "max_depth": 4
    },
    "neural_network": {
        "hidden_layers": [32, 16, 8],
        "activation": "relu",
        "optimizer": "adam",
        "learning_rate": 0.001,
        "epochs": 200,
        "batch_size": 2
    },
    "output_files": {
        "h5": os.path.basename(H5_PATH),
        "pkl": os.path.basename(PKL_PATH),
        "yaml": os.path.basename(YAML_PATH),
        "json": os.path.basename(JSON_PATH)
    }
}

with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:
    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("\nYAML configuration saved:")
print(YAML_PATH)

# ============================================================
# STEP 17: SAVE JSON METADATA
# ============================================================

metadata = {
    "project_name": config["project_name"],
    "created_at": datetime.now().isoformat(),
    "dataset_information": {
        "records": int(len(df)),
        "columns": int(df.shape[1]),
        "training_records": int(len(X_train)),
        "testing_records": int(len(X_test)),
        "features": feature_columns,
        "target": "PaidupCapital"
    },
    "capital_statistics": {
        "average_authorized_capital": (
            float(df["AuthorizedCapital"].mean())
        ),
        "average_paidup_capital": (
            float(df["PaidupCapital"].mean())
        ),
        "maximum_paidup_capital": (
            float(df["PaidupCapital"].max())
        ),
        "minimum_paidup_capital": (
            float(df["PaidupCapital"].min())
        ),
        "average_capital_utilization": (
            float(df["CapitalUtilization"].mean())
        )
    },
    "model_performance": {
        "random_forest": {
            "mae": float(rf_mae),
            "rmse": float(rf_rmse),
            "r2": (
                float(rf_r2)
                if rf_r2 is not None
                else None
            ),
            "cross_validation_mae": float(
                cv_mae_scores.mean()
            )
        },
        "neural_network": {
            "mae": float(nn_mae),
            "rmse": float(nn_rmse),
            "r2": (
                float(nn_r2)
                if nn_r2 is not None
                else None
            ),
            "epochs_trained": len(
                history.history["loss"]
            )
        }
    },
    "limitations": (
        "The dataset is very small. Model metrics "
        "are exploratory and should not be treated "
        "as reliable out-of-sample performance."
    )
}

with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        metadata,
        file,
        indent=4,
        allow_nan=False
    )

print("\nJSON metadata saved:")
print(JSON_PATH)

# ============================================================
# STEP 18: TEST SAVED MODELS
# ============================================================

print("\n" + "=" * 60)
print("VERIFYING SAVED FILES")
print("=" * 60)

loaded_bundle = joblib.load(PKL_PATH)

loaded_rf = loaded_bundle[
    "random_forest_pipeline"
]

loaded_nn = tf.keras.models.load_model(
    H5_PATH,
    compile=False
)

loaded_preprocessor = loaded_bundle[
    "neural_network_preprocessor"
]

loaded_target_scaler = loaded_bundle[
    "neural_network_target_scaler"
]

sample = X_test.iloc[[0]]

rf_sample_prediction = loaded_rf.predict(
    sample
)[0]

nn_sample_features = loaded_preprocessor.transform(
    sample
)

nn_sample_prediction_scaled = loaded_nn.predict(
    np.asarray(
        nn_sample_features,
        dtype=np.float32
    ),
    verbose=0
)

nn_sample_prediction = (
    loaded_target_scaler.inverse_transform(
        nn_sample_prediction_scaled
    )[0][0]
)

print("\nActual paid-up capital:")
print(float(y_test.iloc[0]))

print("\nRandom Forest prediction:")
print(float(rf_sample_prediction))

print("\nNeural Network prediction:")
print(float(nn_sample_prediction))

# ============================================================
# STEP 19: DISPLAY FINAL OUTPUT FILES
# ============================================================

print("\n" + "=" * 60)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 60)

for filename in [
    H5_PATH,
    PKL_PATH,
    YAML_PATH,
    JSON_PATH
]:
    if os.path.exists(filename):
        size_kb = os.path.getsize(filename) / 1024
        print(
            f"CREATED: {os.path.basename(filename)} "
            f"({size_kb:.2f} KB)"
        )
    else:
        print(
            f"FAILED: {os.path.basename(filename)}"
        )

print("\nAll outputs saved in:")
print(PROJECT_DIR)
```
