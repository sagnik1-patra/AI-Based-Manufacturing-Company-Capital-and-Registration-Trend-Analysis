```python
import os
import json
import yaml
import joblib
import warnings
import random
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ==============================================================
# STEP 1: PROJECT PATHS
# ==============================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Manufacturing Company Capital and Registration Trend Analysis"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "4dbe5667-7b6b-41d7-82af-211562424d9a.csv"
)

os.makedirs(PROJECT_DIR, exist_ok=True)

SEED = 42
np.random.seed(SEED)
random.seed(SEED)

sns.set_theme(style="whitegrid")

def save_path(filename):
    return os.path.join(
        PROJECT_DIR,
        "ais_" + filename
    )

# ==============================================================
# STEP 2: LOAD DATASET
# ==============================================================

if not os.path.isfile(CSV_PATH):
    raise FileNotFoundError(
        f"Dataset not found: {CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH)
df.columns = df.columns.str.strip()

print("=" * 65)
print("AIS MANUFACTURING CAPITAL ANALYSIS")
print("=" * 65)

print("\nDataset Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst Five Records:")
print(df.head())

# ==============================================================
# STEP 3: DATA CLEANING
# ==============================================================

required = [
    "AuthorizedCapital",
    "PaidupCapital",
    "CompanyRegistrationdate_date"
]

for col in required:
    if col not in df.columns:
        raise ValueError(
            f"Missing required column: {col}"
        )

df["AuthorizedCapital"] = pd.to_numeric(
    df["AuthorizedCapital"],
    errors="coerce"
)

df["PaidupCapital"] = pd.to_numeric(
    df["PaidupCapital"],
    errors="coerce"
)

df["CompanyRegistrationdate_date"] = pd.to_datetime(
    df["CompanyRegistrationdate_date"],
    errors="coerce"
)

df = df.dropna(
    subset=["PaidupCapital"]
).copy()

if len(df) < 6:
    raise ValueError(
        "At least six valid rows are required."
    )

# ==============================================================
# STEP 4: FEATURE ENGINEERING
# ==============================================================

df["RegistrationYear"] = (
    df["CompanyRegistrationdate_date"].dt.year
)

df["RegistrationMonth"] = (
    df["CompanyRegistrationdate_date"].dt.month
)

df["RegistrationDay"] = (
    df["CompanyRegistrationdate_date"].dt.day
)

# Descriptive feature only.
# Excluded from training to avoid target leakage.

df["CapitalUtilization"] = np.where(
    df["AuthorizedCapital"] > 0,
    df["PaidupCapital"] /
    df["AuthorizedCapital"] * 100,
    np.nan
)

numeric_features = [
    "AuthorizedCapital",
    "RegistrationYear",
    "RegistrationMonth",
    "RegistrationDay"
]

categorical_candidates = [
    "CompanyROCcode",
    "CompanyCategory",
    "CompanySubCategory",
    "CompanyClass",
    "Listingstatus",
    "CompanyStatus",
    "CompanyStateCode",
    "CompanyIndian/Foreign Company",
    "CompanyIndustrialClassification"
]

categorical_features = [
    col for col in categorical_candidates
    if col in df.columns
]

feature_columns = (
    numeric_features + categorical_features
)

X = df[feature_columns].copy()
y = df["PaidupCapital"].astype(float)

for col in categorical_features:
    X[col] = (
        X[col]
        .astype("string")
        .fillna("Missing")
        .astype(str)
    )

# ==============================================================
# STEP 5: TRAIN-TEST SPLIT
# ==============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED
)

print("\nTraining Records:", len(X_train))
print("Testing Records:", len(X_test))

# ==============================================================
# STEP 6: PREPROCESSING PIPELINE
# ==============================================================

def create_preprocessor():

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ])

    try:
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    except TypeError:
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "encoder",
            encoder
        )
    ])

    return ColumnTransformer([
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ])


def build_pipeline(params):

    return Pipeline([
        (
            "preprocessor",
            create_preprocessor()
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=int(params[0]),
                max_depth=int(params[1]),
                min_samples_split=int(params[2]),
                min_samples_leaf=int(params[3]),
                random_state=SEED,
                n_jobs=-1
            )
        )
    ])

# ==============================================================
# STEP 7: AIS PARAMETER SEARCH SPACE
# ==============================================================

# Antibody represents Random Forest hyperparameters:
# [n_estimators, max_depth, min_samples_split,
#  min_samples_leaf]

BOUNDS = np.array([
    [20, 150],
    [2, 12],
    [2, 6],
    [1, 3]
], dtype=int)

PARAMETER_NAMES = [
    "n_estimators",
    "max_depth",
    "min_samples_split",
    "min_samples_leaf"
]

rng = np.random.default_rng(SEED)

def random_antibody():

    return np.array([
        rng.integers(low, high + 1)
        for low, high in BOUNDS
    ], dtype=int)


def clip_antibody(antibody):

    return np.clip(
        np.rint(antibody).astype(int),
        BOUNDS[:, 0],
        BOUNDS[:, 1]
    )

# ==============================================================
# STEP 8: FITNESS FUNCTION
# ==============================================================

# Important:
# AIS searches using only training data.
# The test set is not used for hyperparameter selection.

cv = KFold(
    n_splits=min(4, len(X_train)),
    shuffle=True,
    random_state=SEED
)

fitness_cache = {}

def fitness(antibody):

    antibody = clip_antibody(antibody)

    key = tuple(
        int(v) for v in antibody
    )

    if key in fitness_cache:
        return fitness_cache[key]

    pipeline = build_pipeline(antibody)

    scores = cross_val_score(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring="neg_mean_absolute_error",
        n_jobs=1
    )

    # Lower MAE is better

    error = float(-np.mean(scores))

    fitness_cache[key] = error

    return error

# ==============================================================
# STEP 9: CLONAL SELECTION ALGORITHM
# ==============================================================

POPULATION_SIZE = 8
GENERATIONS = 12
TOP_SELECTION = 4
CLONES_PER_ANTIBODY = 3
RANDOM_REPLACEMENTS = 2

population = [
    random_antibody()
    for _ in range(POPULATION_SIZE)
]

best_antibody = None
best_fitness = float("inf")

convergence_history = []
optimization_records = []

def hypermutate(antibody, rank):

    mutated = antibody.copy()

    # Better-ranked antibodies receive smaller mutations.

    mutation_probability = min(
        0.90,
        0.15 + 0.15 * rank
    )

    for i in range(len(mutated)):

        if rng.random() < mutation_probability:

            span = (
                BOUNDS[i, 1] -
                BOUNDS[i, 0]
            )

            scale = max(
                1,
                int(span * (0.05 + 0.05 * rank))
            )

            mutated[i] += rng.integers(
                -scale,
                scale + 1
            )

    return clip_antibody(mutated)


print("\nStarting AIS optimization...\n")

for generation in range(GENERATIONS):

    evaluated = [
        (fitness(a), a.copy())
        for a in population
    ]

    evaluated.sort(
        key=lambda item: item[0]
    )

    current_error, current_best = evaluated[0]

    if current_error < best_fitness:

        best_fitness = current_error
        best_antibody = current_best.copy()

    selected = evaluated[:TOP_SELECTION]

    clones = []

    for rank, (_, antibody) in enumerate(selected):

        for _ in range(CLONES_PER_ANTIBODY):

            clone = hypermutate(
                antibody,
                rank
            )

            clones.append(clone)

    candidates = (
        [a for _, a in evaluated] +
        clones +
        [
            random_antibody()
            for _ in range(RANDOM_REPLACEMENTS)
        ]
    )

    unique_candidates = {}

    for antibody in candidates:

        key = tuple(antibody.tolist())

        unique_candidates[key] = antibody

    ranked_candidates = sorted(
        unique_candidates.values(),
        key=fitness
    )

    population = ranked_candidates[
        :POPULATION_SIZE
    ]

    # Maintain population size

    while len(population) < POPULATION_SIZE:

        population.append(
            random_antibody()
        )

    # Update best from the new population too

    new_best = population[0]
    new_error = fitness(new_best)

    if new_error < best_fitness:

        best_fitness = new_error
        best_antibody = new_best.copy()

    convergence_history.append(
        best_fitness
    )

    optimization_records.append({
        "Generation": generation + 1,
        "Best_CV_MAE": best_fitness,
        **{
            name: int(value)
            for name, value in zip(
                PARAMETER_NAMES,
                best_antibody
            )
        }
    })

    print(
        f"Generation {generation + 1:02d} | "
        f"Best CV MAE: {best_fitness:.4f} | "
        f"Parameters: {best_antibody.tolist()}"
    )

print("\nAIS Optimization Completed.")

best_params = {
    name: int(value)
    for name, value in zip(
        PARAMETER_NAMES,
        best_antibody
    )
}

print("\nBest Parameters:")
print(best_params)

# ==============================================================
# STEP 10: TRAIN BASELINE AND AIS MODELS
# ==============================================================

baseline_params = [100, 4, 2, 1]

baseline_model = build_pipeline(
    baseline_params
)

ais_model = build_pipeline(
    best_antibody
)

baseline_model.fit(
    X_train,
    y_train
)

ais_model.fit(
    X_train,
    y_train
)

baseline_predictions = baseline_model.predict(
    X_test
)

ais_predictions = ais_model.predict(
    X_test
)

actual_values = y_test.to_numpy()

# ==============================================================
# STEP 11: EVALUATION METRICS
# ==============================================================

def evaluate_model(actual, predicted):

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(mse)

    r2 = (
        r2_score(actual, predicted)
        if len(actual) >= 2
        else np.nan
    )

    denominator = np.mean(
        np.abs(actual)
    )

    nmae = (
        mae / denominator
        if denominator > 0
        else np.nan
    )

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2": float(r2),
        "NMAE": float(nmae)
    }


baseline_metrics = evaluate_model(
    actual_values,
    baseline_predictions
)

ais_metrics = evaluate_model(
    actual_values,
    ais_predictions
)

print("\nBaseline Metrics:")
print(baseline_metrics)

print("\nAIS Metrics:")
print(ais_metrics)

# ==============================================================
# STEP 12: SAVE AIS RESULTS CSV
# ==============================================================

results_df = pd.DataFrame([
    {
        "Model": "Baseline Random Forest",
        **baseline_metrics
    },
    {
        "Model": "AIS Optimized Random Forest",
        **ais_metrics
    }
])

results_df.to_csv(
    save_path("results.csv"),
    index=False
)

print("\nResults:")
print(results_df)

# ==============================================================
# STEP 13: SAVE AIS PREDICTION CSV
# ==============================================================

prediction_df = pd.DataFrame({
    "Original_Row_Index": X_test.index,
    "Actual_PaidupCapital": actual_values,
    "Baseline_Prediction": baseline_predictions,
    "AIS_Prediction": ais_predictions
})

prediction_df["Baseline_Absolute_Error"] = np.abs(
    prediction_df["Actual_PaidupCapital"] -
    prediction_df["Baseline_Prediction"]
)

prediction_df["AIS_Absolute_Error"] = np.abs(
    prediction_df["Actual_PaidupCapital"] -
    prediction_df["AIS_Prediction"]
)

prediction_df = prediction_df.sort_values(
    "Original_Row_Index"
).reset_index(drop=True)

prediction_df.to_csv(
    save_path("prediction.csv"),
    index=False
)

print("\nPredictions:")
print(prediction_df)

# ==============================================================
# STEP 14: AIS ACCURACY / PERFORMANCE GRAPH
# ==============================================================

# This is a regression problem.
# R2 and NMAE are shown, not classification accuracy.

fig, axes = plt.subplots(
    1, 2,
    figsize=(14, 5)
)

models = results_df["Model"]

bars1 = axes[0].bar(
    models,
    results_df["R2"],
    color=["steelblue", "seagreen"]
)

axes[0].axhline(
    0,
    color="black",
    linewidth=1
)

axes[0].set_title(
    "Baseline vs AIS: R2 Score"
)

axes[0].set_ylabel("R2 Score")

axes[0].tick_params(
    axis="x",
    labelrotation=12
)

axes[0].bar_label(
    bars1,
    fmt="%.3f"
)

bars2 = axes[1].bar(
    models,
    results_df["NMAE"],
    color=["darkorange", "mediumpurple"]
)

axes[1].set_title(
    "Baseline vs AIS: Normalized MAE"
)

axes[1].set_ylabel(
    "NMAE (Lower is Better)"
)

axes[1].tick_params(
    axis="x",
    labelrotation=12
)

axes[1].bar_label(
    bars2,
    fmt="%.3f"
)

fig.suptitle(
    "AIS Model Performance Analysis",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    save_path("accuracy_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

# ==============================================================
# STEP 15: AIS CORRELATION HEATMAP
# ==============================================================

heatmap_columns = [
    "AuthorizedCapital",
    "PaidupCapital",
    "RegistrationYear",
    "RegistrationMonth",
    "RegistrationDay",
    "CapitalUtilization"
]

correlation = df[
    heatmap_columns
].corr()

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5
)

plt.title(
    "AIS Project: Manufacturing Capital Correlation",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    save_path("heatmap.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# STEP 16: AIS COMPARISON GRAPH
# ==============================================================

comparison_df = results_df.melt(
    id_vars="Model",
    value_vars=["MAE", "RMSE"],
    var_name="Metric",
    value_name="Error"
)

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Metric",
    y="Error",
    hue="Model",
    palette="Set2"
)

plt.title(
    "Baseline Random Forest vs AIS Optimization",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Metric")
plt.ylabel("Error (Capital Units)")

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.2f",
        padding=3
    )

plt.tight_layout()

plt.savefig(
    save_path("comparison_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# STEP 17: AIS RESULT GRAPH
# ==============================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(14, 5)
)

all_values = np.concatenate([
    actual_values,
    baseline_predictions,
    ais_predictions
])

minimum = np.min(all_values)
maximum = np.max(all_values)

axes[0].scatter(
    actual_values,
    baseline_predictions,
    color="royalblue",
    s=80
)

axes[1].scatter(
    actual_values,
    ais_predictions,
    color="seagreen",
    s=80
)

for ax in axes:

    ax.plot(
        [minimum, maximum],
        [minimum, maximum],
        "k--",
        label="Perfect Prediction"
    )

    ax.set_xlabel("Actual Paid-up Capital")
    ax.set_ylabel("Predicted Paid-up Capital")
    ax.legend()
    ax.grid(alpha=0.3)

axes[0].set_title(
    "Baseline Random Forest"
)

axes[1].set_title(
    "AIS Optimized Random Forest"
)

fig.suptitle(
    "AIS Prediction Result Analysis",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    save_path("result_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

# ==============================================================
# STEP 18: AIS PREDICTION GRAPH
# ==============================================================

positions = np.arange(
    len(prediction_df)
)

plt.figure(figsize=(12, 6))

plt.plot(
    positions,
    prediction_df["Actual_PaidupCapital"],
    marker="o",
    linewidth=2.5,
    label="Actual Capital"
)

plt.plot(
    positions,
    prediction_df["Baseline_Prediction"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Baseline Prediction"
)

plt.plot(
    positions,
    prediction_df["AIS_Prediction"],
    marker="^",
    linestyle="--",
    linewidth=2,
    label="AIS Prediction"
)

plt.xticks(
    positions,
    prediction_df["Original_Row_Index"].astype(str)
)

plt.title(
    "AIS: Actual vs Predicted Paid-up Capital",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Original Dataset Row Index")
plt.ylabel("Paid-up Capital")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    save_path("prediction_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# STEP 19: AIS CONVERGENCE GRAPH
# ==============================================================

optimization_df = pd.DataFrame(
    optimization_records
)

optimization_df.to_csv(
    save_path("optimization_history.csv"),
    index=False
)

plt.figure(figsize=(11, 6))

plt.plot(
    range(1, GENERATIONS + 1),
    convergence_history,
    marker="o",
    linewidth=2.5,
    color="purple"
)

plt.title(
    "AIS Clonal Selection Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Generation")
plt.ylabel("Best Cross-Validation MAE")

plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    save_path("convergence_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# STEP 20: SAVE OPTIMIZED MODEL
# ==============================================================

model_bundle = {
    "model": ais_model,
    "feature_columns": feature_columns,
    "numeric_features": numeric_features,
    "categorical_features": categorical_features,
    "target_column": "PaidupCapital",
    "best_parameters": best_params,
    "random_state": SEED
}

joblib.dump(
    model_bundle,
    save_path("model.pkl")
)

print("\nAIS model saved successfully.")

# ==============================================================
# STEP 21: SAVE AIS YAML CONFIGURATION
# ==============================================================

config = {
    "project": (
        "AI-Based Manufacturing Company Capital "
        "and Registration Trend Analysis"
    ),
    "optimization_algorithm": "AIS_CLONALG",
    "prediction_model": "RandomForestRegressor",
    "dataset_path": CSV_PATH,
    "target": "PaidupCapital",
    "random_state": SEED,
    "test_size": 0.20,
    "ais_settings": {
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "top_selection": TOP_SELECTION,
        "clones_per_antibody": CLONES_PER_ANTIBODY,
        "random_replacements": RANDOM_REPLACEMENTS
    },
    "best_hyperparameters": best_params,
    "best_cross_validation_mae": float(
        best_fitness
    )
}

with open(
    save_path("config.yaml"),
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

# ==============================================================
# STEP 22: SAVE AIS JSON METADATA
# ==============================================================

metadata = {
    "project_name": config["project"],
    "algorithm": "Artificial Immune System",
    "method": "Clonal Selection",
    "created_at": datetime.now().isoformat(),
    "dataset_records": int(len(df)),
    "training_records": int(len(X_train)),
    "testing_records": int(len(X_test)),
    "features": feature_columns,
    "target": "PaidupCapital",
    "best_parameters": best_params,
    "best_cv_mae": float(best_fitness),
    "baseline_metrics": baseline_metrics,
    "ais_metrics": ais_metrics,
    "total_unique_fitness_evaluations": int(
        len(fitness_cache)
    ),
    "limitations": (
        "Small dataset. Optimization and evaluation "
        "results are exploratory and may be unstable."
    )
}

with open(
    save_path("metadata.json"),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4,
        allow_nan=False
    )

# ==============================================================
# STEP 23: FINAL OUTPUT VERIFICATION
# ==============================================================

print("\n" + "=" * 65)
print("AIS PROJECT COMPLETED")
print("=" * 65)

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "results.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "convergence_graph.png",
    "optimization_history.csv",
    "model.pkl",
    "config.yaml",
    "metadata.json"
]

for filename in output_files:

    path = save_path(filename)

    if os.path.isfile(path):

        size_kb = os.path.getsize(path) / 1024

        print(
            f"[SAVED] ais_{filename} "
            f"({size_kb:.2f} KB)"
        )

    else:
        print(
            f"[MISSING] ais_{filename}"
        )

print("\nBest AIS Hyperparameters:")
print(best_params)

print("\nBest AIS CV MAE:", best_fitness)

print("\nBaseline Test MAE:", baseline_metrics["MAE"])
print("AIS Test MAE:", ais_metrics["MAE"])

print("\nBaseline Test R2:", baseline_metrics["R2"])
print("AIS Test R2:", ais_metrics["R2"])

print("\nAll AIS results saved in:")
print(PROJECT_DIR)
```
