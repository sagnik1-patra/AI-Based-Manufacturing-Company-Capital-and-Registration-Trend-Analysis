```python
import os
import json
import yaml
import joblib
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder
)

from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ==============================================================
# STEP 1: PROJECT DIRECTORY
# ==============================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Manufacturing Company Capital and Registration Trend Analysis"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "4dbe5667-7b6b-41d7-82af-211562424d9a.csv"
)

os.makedirs(PROJECT_DIR, exist_ok=True)

SEED = 42
rng = np.random.default_rng(SEED)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (11, 6)
plt.rcParams["figure.dpi"] = 120

def save_path(filename):
    return os.path.join(
        PROJECT_DIR,
        "pso_" + filename
    )

# ==============================================================
# STEP 2: LOAD DATASET
# ==============================================================

if not os.path.isfile(CSV_PATH):
    raise FileNotFoundError(
        f"CSV file not found: {CSV_PATH}"
    )

df = pd.read_csv(CSV_PATH)

df.columns = df.columns.str.strip()

print("=" * 65)
print("PSO MANUFACTURING CAPITAL ANALYSIS")
print("=" * 65)

print("\nDataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst Five Records:")
print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

# ==============================================================
# STEP 3: DATA CLEANING
# ==============================================================

required_columns = [
    "AuthorizedCapital",
    "PaidupCapital",
    "CompanyRegistrationdate_date"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

df["AuthorizedCapital"] = pd.to_numeric(
    df["AuthorizedCapital"],
    errors="coerce"
)

df["PaidupCapital"] = pd.to_numeric(
    df["PaidupCapital"],
    errors="coerce"
)

df["CompanyRegistrationdate_date"] = pd.to_datetime(
    df["CompanyRegistrationdate_date"],
    errors="coerce"
)

df = df.dropna(
    subset=["PaidupCapital"]
).copy()

if len(df) < 6:
    raise ValueError(
        "At least six valid records are required."
    )

# ==============================================================
# STEP 4: FEATURE ENGINEERING
# ==============================================================

df["RegistrationYear"] = (
    df["CompanyRegistrationdate_date"].dt.year
)

df["RegistrationMonth"] = (
    df["CompanyRegistrationdate_date"].dt.month
)

df["RegistrationDay"] = (
    df["CompanyRegistrationdate_date"].dt.day
)

# Capital utilization is descriptive only.
# It uses the target and must not enter the model.

df["CapitalUtilization"] = np.where(
    df["AuthorizedCapital"] > 0,
    (
        df["PaidupCapital"] /
        df["AuthorizedCapital"]
    ) * 100,
    np.nan
)

numeric_features = [
    "AuthorizedCapital",
    "RegistrationYear",
    "RegistrationMonth",
    "RegistrationDay"
]

categorical_candidates = [
    "CompanyROCcode",
    "CompanyCategory",
    "CompanySubCategory",
    "CompanyClass",
    "Listingstatus",
    "CompanyStatus",
    "CompanyStateCode",
    "CompanyIndian/Foreign Company",
    "CompanyIndustrialClassification"
]

categorical_features = [
    col for col in categorical_candidates
    if col in df.columns
]

feature_columns = (
    numeric_features + categorical_features
)

X = df[feature_columns].copy()

y = df["PaidupCapital"].astype(float)

for col in categorical_features:
    X[col] = (
        X[col]
        .astype("string")
        .fillna("Missing")
        .astype(str)
    )

print("\nSelected Features:")
print(feature_columns)

print("\nTarget: PaidupCapital")

# ==============================================================
# STEP 5: TRAIN-TEST SPLIT
# ==============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED
)

print("\nTraining Records:", len(X_train))
print("Testing Records:", len(X_test))

# ==============================================================
# STEP 6: CREATE PREPROCESSING PIPELINE
# ==============================================================

def create_preprocessor():

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ])

    try:
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )
    except TypeError:
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),
        (
            "encoder",
            encoder
        )
    ])

    return ColumnTransformer([
        (
            "numeric",
            numeric_pipeline,
            numeric_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ])


def build_pipeline(params):

    return Pipeline([
        (
            "preprocessor",
            create_preprocessor()
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=int(params[0]),
                max_depth=int(params[1]),
                min_samples_split=int(params[2]),
                min_samples_leaf=int(params[3]),
                random_state=SEED,
                n_jobs=1
            )
        )
    ])

# ==============================================================
# STEP 7: DEFINE PSO SEARCH SPACE
# ==============================================================

# Each particle represents:
#
# [n_estimators,
#  max_depth,
#  min_samples_split,
#  min_samples_leaf]

PARAMETER_NAMES = [
    "n_estimators",
    "max_depth",
    "min_samples_split",
    "min_samples_leaf"
]

LOWER_BOUNDS = np.array([
    20, 2, 2, 1
], dtype=float)

UPPER_BOUNDS = np.array([
    150, 12, 6, 3
], dtype=float)

DIMENSIONS = len(PARAMETER_NAMES)

def decode_particle(position):

    values = np.rint(position).astype(int)

    return np.clip(
        values,
        LOWER_BOUNDS.astype(int),
        UPPER_BOUNDS.astype(int)
    )

# ==============================================================
# STEP 8: PSO FITNESS FUNCTION
# ==============================================================

# Minimize cross-validation MAE.
# Only training data is used during optimization.

cv = KFold(
    n_splits=min(4, len(X_train)),
    shuffle=True,
    random_state=SEED
)

fitness_cache = {}

def fitness(position):

    params = decode_particle(position)

    key = tuple(
        int(value)
        for value in params
    )

    if key in fitness_cache:
        return fitness_cache[key]

    pipeline = build_pipeline(params)

    scores = cross_val_score(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring="neg_mean_absolute_error",
        n_jobs=1
    )

    error = float(
        -np.mean(scores)
    )

    fitness_cache[key] = error

    return error

# ==============================================================
# STEP 9: INITIALIZE PARTICLE SWARM
# ==============================================================

NUM_PARTICLES = 8
MAX_ITERATIONS = 12

INERTIA_WEIGHT = 0.7
COGNITIVE_COEFFICIENT = 1.5
SOCIAL_COEFFICIENT = 1.5

positions = rng.uniform(
    LOWER_BOUNDS,
    UPPER_BOUNDS,
    size=(NUM_PARTICLES, DIMENSIONS)
)

velocities = np.zeros(
    (NUM_PARTICLES, DIMENSIONS)
)

personal_best_positions = positions.copy()

personal_best_scores = np.array([
    fitness(position)
    for position in positions
])

best_index = np.argmin(
    personal_best_scores
)

global_best_position = (
    personal_best_positions[best_index].copy()
)

global_best_score = float(
    personal_best_scores[best_index]
)

convergence_history = []
optimization_records = []

# ==============================================================
# STEP 10: PARTICLE SWARM OPTIMIZATION
# ==============================================================

print("\nStarting PSO Optimization...\n")

for iteration in range(MAX_ITERATIONS):

    for particle in range(NUM_PARTICLES):

        r1 = rng.random(DIMENSIONS)
        r2 = rng.random(DIMENSIONS)

        # Velocity update

        velocities[particle] = (
            INERTIA_WEIGHT * velocities[particle]
            +
            COGNITIVE_COEFFICIENT
            * r1
            * (
                personal_best_positions[particle]
                - positions[particle]
            )
            +
            SOCIAL_COEFFICIENT
            * r2
            * (
                global_best_position
                - positions[particle]
            )
        )

        # Position update

        positions[particle] = (
            positions[particle]
            + velocities[particle]
        )

        # Boundary handling

        positions[particle] = np.clip(
            positions[particle],
            LOWER_BOUNDS,
            UPPER_BOUNDS
        )

        current_score = fitness(
            positions[particle]
        )

        # Update personal best

        if (
            current_score
            < personal_best_scores[particle]
        ):

            personal_best_scores[particle] = (
                current_score
            )

            personal_best_positions[particle] = (
                positions[particle].copy()
            )

            # Update global best

            if current_score < global_best_score:

                global_best_score = float(
                    current_score
                )

                global_best_position = (
                    positions[particle].copy()
                )

    convergence_history.append(
        global_best_score
    )

    best_params_iteration = decode_particle(
        global_best_position
    )

    optimization_records.append({
        "Iteration": iteration + 1,
        "Best_CV_MAE": global_best_score,
        **{
            name: int(value)
            for name, value in zip(
                PARAMETER_NAMES,
                best_params_iteration
            )
        }
    })

    print(
        f"Iteration {iteration + 1:02d} | "
        f"Best CV MAE: {global_best_score:.4f} | "
        f"Parameters: {best_params_iteration.tolist()}"
    )

print("\nPSO Optimization Completed.")

# ==============================================================
# STEP 11: GET BEST HYPERPARAMETERS
# ==============================================================

best_particle = decode_particle(
    global_best_position
)

best_params = {
    name: int(value)
    for name, value in zip(
        PARAMETER_NAMES,
        best_particle
    )
}

print("\nBest PSO Parameters:")
print(best_params)

print("\nBest Cross-Validation MAE:")
print(global_best_score)

# ==============================================================
# STEP 12: TRAIN BASELINE AND PSO MODELS
# ==============================================================

baseline_params = [
    100, 4, 2, 1
]

baseline_model = build_pipeline(
    baseline_params
)

pso_model = build_pipeline(
    best_particle
)

baseline_model.fit(
    X_train,
    y_train
)

pso_model.fit(
    X_train,
    y_train
)

baseline_predictions = baseline_model.predict(
    X_test
)

pso_predictions = pso_model.predict(
    X_test
)

actual_values = y_test.to_numpy()

print("\nModel training completed.")

# ==============================================================
# STEP 13: EVALUATION METRICS
# ==============================================================

def evaluate_model(actual, predicted):

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(mse)

    r2 = (
        r2_score(actual, predicted)
        if len(actual) >= 2
        else np.nan
    )

    denominator = np.mean(
        np.abs(actual)
    )

    nmae = (
        mae / denominator
        if denominator > 0
        else np.nan
    )

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2": float(r2),
        "NMAE": float(nmae)
    }


baseline_metrics = evaluate_model(
    actual_values,
    baseline_predictions
)

pso_metrics = evaluate_model(
    actual_values,
    pso_predictions
)

print("\n" + "=" * 65)
print("MODEL PERFORMANCE RESULTS")
print("=" * 65)

print("\nBaseline Random Forest:")
print(baseline_metrics)

print("\nPSO Optimized Random Forest:")
print(pso_metrics)

# ==============================================================
# STEP 14: SAVE RESULTS CSV
# ==============================================================

results_df = pd.DataFrame([
    {
        "Model": "Baseline Random Forest",
        **baseline_metrics
    },
    {
        "Model": "PSO Optimized Random Forest",
        **pso_metrics
    }
])

results_df.to_csv(
    save_path("results.csv"),
    index=False
)

print("\nResults CSV:")
print(results_df)

# ==============================================================
# STEP 15: SAVE PREDICTION CSV
# ==============================================================

prediction_df = pd.DataFrame({
    "Original_Row_Index": X_test.index,
    "Actual_PaidupCapital": actual_values,
    "Baseline_Prediction": baseline_predictions,
    "PSO_Prediction": pso_predictions
})

prediction_df["Baseline_Absolute_Error"] = np.abs(
    prediction_df["Actual_PaidupCapital"]
    - prediction_df["Baseline_Prediction"]
)

prediction_df["PSO_Absolute_Error"] = np.abs(
    prediction_df["Actual_PaidupCapital"]
    - prediction_df["PSO_Prediction"]
)

prediction_df = prediction_df.sort_values(
    "Original_Row_Index"
).reset_index(drop=True)

prediction_df.to_csv(
    save_path("prediction.csv"),
    index=False
)

print("\nPrediction CSV:")
print(prediction_df)

# ==============================================================
# STEP 16: PSO ACCURACY / PERFORMANCE GRAPH
# ==============================================================

# Regression uses R2 and normalized MAE,
# not classification accuracy.

fig, axes = plt.subplots(
    1,
    2,
    figsize=(14, 5)
)

model_names = results_df["Model"]

bars1 = axes[0].bar(
    model_names,
    results_df["R2"],
    color=["steelblue", "seagreen"]
)

axes[0].axhline(
    0,
    color="black",
    linewidth=1
)

axes[0].set_title(
    "Baseline vs PSO: R2 Score"
)

axes[0].set_ylabel("R2 Score")

axes[0].tick_params(
    axis="x",
    labelrotation=12
)

axes[0].bar_label(
    bars1,
    fmt="%.3f"
)

bars2 = axes[1].bar(
    model_names,
    results_df["NMAE"],
    color=["darkorange", "mediumpurple"]
)

axes[1].set_title(
    "Baseline vs PSO: Normalized MAE"
)

axes[1].set_ylabel(
    "NMAE (Lower is Better)"
)

axes[1].tick_params(
    axis="x",
    labelrotation=12
)

axes[1].bar_label(
    bars2,
    fmt="%.3f"
)

fig.suptitle(
    "PSO Model Performance Analysis",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    save_path("accuracy_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

# ==============================================================
# STEP 17: PSO CORRELATION HEATMAP
# ==============================================================

heatmap_columns = [
    "AuthorizedCapital",
    "PaidupCapital",
    "RegistrationYear",
    "RegistrationMonth",
    "RegistrationDay",
    "CapitalUtilization"
]

correlation_matrix = df[
    heatmap_columns
].corr()

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5
)

plt.title(
    "PSO Project: Manufacturing Capital Correlation",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    save_path("heatmap.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# STEP 18: PSO MODEL COMPARISON GRAPH
# ==============================================================

comparison_df = results_df.melt(
    id_vars="Model",
    value_vars=["MAE", "RMSE"],
    var_name="Metric",
    value_name="Error"
)

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Metric",
    y="Error",
    hue="Model",
    palette="Set2"
)

plt.title(
    "Baseline Random Forest vs PSO Optimization",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Evaluation Metric")
plt.ylabel("Error (Capital Units)")

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.2f",
        padding=3
    )

plt.tight_layout()

plt.savefig(
    save_path("comparison_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# STEP 19: PSO RESULT GRAPH
# ==============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(14, 5)
)

all_values = np.concatenate([
    actual_values,
    baseline_predictions,
    pso_predictions
])

minimum = np.min(all_values)
maximum = np.max(all_values)

axes[0].scatter(
    actual_values,
    baseline_predictions,
    color="royalblue",
    s=80,
    label="Baseline Predictions"
)

axes[1].scatter(
    actual_values,
    pso_predictions,
    color="seagreen",
    s=80,
    label="PSO Predictions"
)

for ax in axes:

    ax.plot(
        [minimum, maximum],
        [minimum, maximum],
        "k--",
        label="Perfect Prediction"
    )

    ax.set_xlabel("Actual Paid-up Capital")
    ax.set_ylabel("Predicted Paid-up Capital")
    ax.legend()
    ax.grid(alpha=0.3)

axes[0].set_title(
    "Baseline Random Forest"
)

axes[1].set_title(
    "PSO Optimized Random Forest"
)

fig.suptitle(
    "PSO Prediction Result Analysis",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    save_path("result_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

# ==============================================================
# STEP 20: PSO PREDICTION GRAPH
# ==============================================================

positions = np.arange(
    len(prediction_df)
)

plt.figure(figsize=(12, 6))

plt.plot(
    positions,
    prediction_df["Actual_PaidupCapital"],
    marker="o",
    linewidth=2.5,
    label="Actual Capital"
)

plt.plot(
    positions,
    prediction_df["Baseline_Prediction"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Baseline Prediction"
)

plt.plot(
    positions,
    prediction_df["PSO_Prediction"],
    marker="^",
    linestyle="--",
    linewidth=2,
    label="PSO Prediction"
)

plt.xticks(
    positions,
    prediction_df["Original_Row_Index"].astype(str)
)

plt.title(
    "PSO: Actual vs Predicted Paid-up Capital",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Original Dataset Row Index")
plt.ylabel("Paid-up Capital")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    save_path("prediction_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# STEP 21: PSO CONVERGENCE GRAPH
# ==============================================================

optimization_df = pd.DataFrame(
    optimization_records
)

optimization_df.to_csv(
    save_path("optimization_history.csv"),
    index=False
)

plt.figure(figsize=(11, 6))

plt.plot(
    range(1, MAX_ITERATIONS + 1),
    convergence_history,
    marker="o",
    linewidth=2.5,
    color="purple"
)

plt.title(
    "PSO Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Iteration")
plt.ylabel("Best Cross-Validation MAE")

plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    save_path("convergence_graph.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

# ==============================================================
# STEP 22: SAVE PSO MODEL
# ==============================================================

model_bundle = {
    "model": pso_model,
    "feature_columns": feature_columns,
    "numeric_features": numeric_features,
    "categorical_features": categorical_features,
    "target_column": "PaidupCapital",
    "best_parameters": best_params,
    "random_state": SEED
}

joblib.dump(
    model_bundle,
    save_path("model.pkl")
)

print("\nPSO optimized model saved.")

# ==============================================================
# STEP 23: SAVE YAML CONFIGURATION
# ==============================================================

config = {
    "project": (
        "AI-Based Manufacturing Company Capital "
        "and Registration Trend Analysis"
    ),
    "optimization_algorithm": (
        "Particle Swarm Optimization"
    ),
    "prediction_model": "RandomForestRegressor",
    "dataset_path": CSV_PATH,
    "target": "PaidupCapital",
    "random_state": SEED,
    "test_size": 0.20,
    "pso_settings": {
        "particles": NUM_PARTICLES,
        "iterations": MAX_ITERATIONS,
        "inertia_weight": INERTIA_WEIGHT,
        "cognitive_coefficient": COGNITIVE_COEFFICIENT,
        "social_coefficient": SOCIAL_COEFFICIENT
    },
    "best_hyperparameters": best_params,
    "best_cross_validation_mae": float(
        global_best_score
    )
}

with open(
    save_path("config.yaml"),
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

# ==============================================================
# STEP 24: SAVE JSON METADATA
# ==============================================================

metadata = {
    "project_name": config["project"],
    "algorithm": "Particle Swarm Optimization",
    "created_at": datetime.now().isoformat(),
    "dataset_records": int(len(df)),
    "training_records": int(len(X_train)),
    "testing_records": int(len(X_test)),
    "features": feature_columns,
    "target": "PaidupCapital",
    "best_parameters": best_params,
    "best_cv_mae": float(global_best_score),
    "baseline_metrics": baseline_metrics,
    "pso_metrics": pso_metrics,
    "unique_fitness_evaluations": int(
        len(fitness_cache)
    ),
    "limitations": (
        "The dataset is very small. Model performance "
        "and optimization results are exploratory."
    )
}

with open(
    save_path("metadata.json"),
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        metadata,
        file,
        indent=4,
        allow_nan=False
    )

# ==============================================================
# STEP 25: VERIFY ALL SAVED OUTPUTS
# ==============================================================

print("\n" + "=" * 65)
print("PSO PROJECT COMPLETED SUCCESSFULLY")
print("=" * 65)

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "results.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png",
    "convergence_graph.png",
    "optimization_history.csv",
    "model.pkl",
    "config.yaml",
    "metadata.json"
]

for filename in output_files:

    path = save_path(filename)

    if os.path.isfile(path):

        size_kb = (
            os.path.getsize(path) / 1024
        )

        print(
            f"[SAVED] pso_{filename} "
            f"({size_kb:.2f} KB)"
        )

    else:
        print(
            f"[MISSING] pso_{filename}"
        )

print("\nBest PSO Hyperparameters:")
print(best_params)

print("\nBest PSO CV MAE:")
print(global_best_score)

print("\nBaseline Test MAE:")
print(baseline_metrics["MAE"])

print("\nPSO Test MAE:")
print(pso_metrics["MAE"])

print("\nBaseline Test R2:")
print(baseline_metrics["R2"])

print("\nPSO Test R2:")
print(pso_metrics["R2"])

print("\nAll PSO results saved in:")
print(PROJECT_DIR)

print("\nExecution completed.")
```
