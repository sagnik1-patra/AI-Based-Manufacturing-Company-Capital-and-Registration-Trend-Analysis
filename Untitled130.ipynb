```python
import os
import warnings
import joblib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# STEP 1: DEFINE PATHS
# ============================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Manufacturing Company Capital and Registration Trend Analysis"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "4dbe5667-7b6b-41d7-82af-211562424d9a.csv"
)

PKL_PATH = os.path.join(
    PROJECT_DIR,
    "manufacturing_model.pkl"
)

H5_PATH = os.path.join(
    PROJECT_DIR,
    "manufacturing_model.h5"
)

os.makedirs(PROJECT_DIR, exist_ok=True)

for path in [CSV_PATH, PKL_PATH, H5_PATH]:
    if not os.path.isfile(path):
        raise FileNotFoundError(
            f"Required file not found: {path}"
        )

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (11, 6)
plt.rcParams["figure.dpi"] = 120

RANDOM_STATE = 42

# ============================================================
# STEP 2: LOAD DATASET
# ============================================================

df = pd.read_csv(CSV_PATH)

df.columns = df.columns.str.strip()

print("=" * 65)
print("DATASET INFORMATION")
print("=" * 65)

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst Five Records:")
print(df.head())

# ============================================================
# STEP 3: DATA PREPROCESSING
# ============================================================

required_columns = [
    "AuthorizedCapital",
    "PaidupCapital",
    "CompanyRegistrationdate_date"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

df["AuthorizedCapital"] = pd.to_numeric(
    df["AuthorizedCapital"],
    errors="coerce"
)

df["PaidupCapital"] = pd.to_numeric(
    df["PaidupCapital"],
    errors="coerce"
)

df["CompanyRegistrationdate_date"] = pd.to_datetime(
    df["CompanyRegistrationdate_date"],
    errors="coerce"
)

df = df.dropna(
    subset=["PaidupCapital"]
).copy()

# Match the previous training code exactly

df["RegistrationYear"] = (
    df["CompanyRegistrationdate_date"].dt.year
)

df["RegistrationMonth"] = (
    df["CompanyRegistrationdate_date"].dt.month
)

df["RegistrationDay"] = (
    df["CompanyRegistrationdate_date"].dt.day
)

df["CapitalUtilization"] = np.where(
    df["AuthorizedCapital"] > 0,
    (
        df["PaidupCapital"] /
        df["AuthorizedCapital"]
    ) * 100,
    np.nan
)

# ============================================================
# STEP 4: LOAD TRAINED MODELS
# ============================================================

model_bundle = joblib.load(PKL_PATH)

rf_model = model_bundle[
    "random_forest_pipeline"
]

nn_preprocessor = model_bundle[
    "neural_network_preprocessor"
]

target_scaler = model_bundle[
    "neural_network_target_scaler"
]

feature_columns = model_bundle[
    "feature_columns"
]

categorical_features = model_bundle[
    "categorical_features"
]

nn_model = tf.keras.models.load_model(
    H5_PATH,
    compile=False
)

print("\nModels loaded successfully.")

# ============================================================
# STEP 5: PREPARE FEATURES
# ============================================================

X = df[feature_columns].copy()

y = df["PaidupCapital"].astype(float)

for col in categorical_features:
    X[col] = (
        X[col]
        .astype("string")
        .fillna("Missing")
        .astype(str)
    )

# Reconstruct original test split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE
)

print("\nTraining Samples:", len(X_train))
print("Testing Samples:", len(X_test))

# ============================================================
# STEP 6: GENERATE MODEL PREDICTIONS
# ============================================================

rf_predictions = rf_model.predict(X_test)

X_test_nn = nn_preprocessor.transform(
    X_test
)

X_test_nn = np.asarray(
    X_test_nn,
    dtype=np.float32
)

nn_scaled_predictions = nn_model.predict(
    X_test_nn,
    verbose=0
)

nn_predictions = (
    target_scaler.inverse_transform(
        nn_scaled_predictions
    ).flatten()
)

actual_values = y_test.to_numpy()

print("\nPredictions generated successfully.")

# ============================================================
# STEP 7: CALCULATE PERFORMANCE METRICS
# ============================================================

def calculate_metrics(actual, predicted):

    mae = mean_absolute_error(
        actual,
        predicted
    )

    mse = mean_squared_error(
        actual,
        predicted
    )

    rmse = np.sqrt(mse)

    r2 = (
        r2_score(actual, predicted)
        if len(actual) >= 2
        else np.nan
    )

    # Normalized MAE relative to mean absolute target
    # Not classification accuracy.

    denominator = np.mean(np.abs(actual))

    nmae = (
        mae / denominator
        if denominator > 0
        else np.nan
    )

    return {
        "MAE": float(mae),
        "MSE": float(mse),
        "RMSE": float(rmse),
        "R2": float(r2),
        "NMAE": float(nmae)
    }


rf_metrics = calculate_metrics(
    actual_values,
    rf_predictions
)

nn_metrics = calculate_metrics(
    actual_values,
    nn_predictions
)

print("\n" + "=" * 65)
print("MODEL PERFORMANCE")
print("=" * 65)

print("\nRandom Forest:")
print(rf_metrics)

print("\nNeural Network:")
print(nn_metrics)

# ============================================================
# STEP 8: SAVE RESULTS CSV
# ============================================================

# Results CSV contains model evaluation metrics

results_df = pd.DataFrame([
    {
        "Model": "Random Forest",
        **rf_metrics
    },
    {
        "Model": "Neural Network",
        **nn_metrics
    }
])

results_path = os.path.join(
    PROJECT_DIR,
    "results.csv"
)

results_df.to_csv(
    results_path,
    index=False
)

print("\nResults CSV:")
print(results_df)

print("\nSaved:", results_path)

# ============================================================
# STEP 9: ACCURACY / PERFORMANCE GRAPH
# ============================================================

# Regression does not have classification accuracy.
# R2 is the coefficient of determination.
# Higher R2 is better, but it can be negative.

fig, axes = plt.subplots(
    1,
    2,
    figsize=(13, 5)
)

model_names = results_df["Model"].tolist()

r2_values = results_df["R2"].to_numpy()

nmae_values = results_df["NMAE"].to_numpy()

bars1 = axes[0].bar(
    model_names,
    r2_values,
    color=["steelblue", "darkorange"]
)

axes[0].axhline(
    y=0,
    color="black",
    linewidth=1
)

axes[0].set_title(
    "Model Performance: R2 Score"
)

axes[0].set_ylabel("R2 Score")

axes[0].bar_label(
    bars1,
    fmt="%.3f",
    padding=3
)

bars2 = axes[1].bar(
    model_names,
    nmae_values,
    color=["seagreen", "mediumpurple"]
)

axes[1].set_title(
    "Normalized Mean Absolute Error"
)

axes[1].set_ylabel(
    "NMAE (Lower is Better)"
)

axes[1].bar_label(
    bars2,
    fmt="%.3f",
    padding=3
)

fig.suptitle(
    "Manufacturing Capital Prediction Performance",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

accuracy_path = os.path.join(
    PROJECT_DIR,
    "accuracy_graph.png"
)

plt.savefig(
    accuracy_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

print("Saved:", accuracy_path)

# ============================================================
# STEP 10: CORRELATION HEATMAP
# ============================================================

heatmap_columns = [
    "AuthorizedCapital",
    "PaidupCapital",
    "RegistrationYear",
    "RegistrationMonth",
    "RegistrationDay",
    "CapitalUtilization"
]

heatmap_df = df[
    heatmap_columns
].copy()

correlation_matrix = heatmap_df.corr(
    method="pearson"
)

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    square=True
)

plt.title(
    "Manufacturing Company Feature Correlation Heatmap",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

heatmap_path = os.path.join(
    PROJECT_DIR,
    "heatmap.png"
)

plt.savefig(
    heatmap_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", heatmap_path)

# ============================================================
# STEP 11: MODEL COMPARISON GRAPH
# ============================================================

# Compare MAE and RMSE in the original capital units.

comparison_df = results_df.melt(
    id_vars="Model",
    value_vars=["MAE", "RMSE"],
    var_name="Metric",
    value_name="Error"
)

plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=comparison_df,
    x="Metric",
    y="Error",
    hue="Model",
    palette="Set2"
)

plt.title(
    "Random Forest vs Neural Network",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Evaluation Metric")
plt.ylabel("Error (Capital Units)")

plt.legend(title="Model")

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.2f",
        padding=3
    )

plt.tight_layout()

comparison_path = os.path.join(
    PROJECT_DIR,
    "comparison_graph.png"
)

plt.savefig(
    comparison_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", comparison_path)

# ============================================================
# STEP 12: RESULT GRAPH
# ============================================================

# Display model errors side by side

fig, axes = plt.subplots(
    1,
    2,
    figsize=(14, 5)
)

axes[0].scatter(
    actual_values,
    rf_predictions,
    color="royalblue",
    s=75,
    label="Random Forest"
)

axes[1].scatter(
    actual_values,
    nn_predictions,
    color="darkorange",
    s=75,
    label="Neural Network"
)

all_values = np.concatenate([
    actual_values,
    rf_predictions,
    nn_predictions
])

minimum = np.min(all_values)
maximum = np.max(all_values)

for ax in axes:

    ax.plot(
        [minimum, maximum],
        [minimum, maximum],
        "k--",
        label="Perfect Prediction"
    )

    ax.set_xlabel("Actual Paid-up Capital")
    ax.set_ylabel("Predicted Paid-up Capital")

    ax.legend()
    ax.grid(alpha=0.3)

axes[0].set_title(
    "Random Forest: Actual vs Predicted"
)

axes[1].set_title(
    "Neural Network: Actual vs Predicted"
)

fig.suptitle(
    "Manufacturing Capital Prediction Results",
    fontsize=15,
    fontweight="bold"
)

plt.tight_layout()

result_graph_path = os.path.join(
    PROJECT_DIR,
    "result_graph.png"
)

plt.savefig(
    result_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

print("Saved:", result_graph_path)

# ============================================================
# STEP 13: GENERATE PREDICTION CSV
# ============================================================

prediction_df = pd.DataFrame({
    "Original_Row_Index": X_test.index,
    "Actual_PaidupCapital": actual_values,
    "RF_Predicted_PaidupCapital": rf_predictions,
    "NN_Predicted_PaidupCapital": nn_predictions
})

prediction_df["RF_Absolute_Error"] = np.abs(
    prediction_df["Actual_PaidupCapital"] -
    prediction_df["RF_Predicted_PaidupCapital"]
)

prediction_df["NN_Absolute_Error"] = np.abs(
    prediction_df["Actual_PaidupCapital"] -
    prediction_df["NN_Predicted_PaidupCapital"]
)

prediction_df = prediction_df.sort_values(
    "Original_Row_Index"
).reset_index(drop=True)

prediction_path = os.path.join(
    PROJECT_DIR,
    "prediction.csv"
)

prediction_df.to_csv(
    prediction_path,
    index=False
)

print("\nPrediction Results:")
print(prediction_df)

print("\nSaved:", prediction_path)

# ============================================================
# STEP 14: PREDICTION GRAPH
# ============================================================

x_positions = np.arange(
    len(prediction_df)
)

plt.figure(figsize=(12, 6))

plt.plot(
    x_positions,
    prediction_df["Actual_PaidupCapital"],
    marker="o",
    linewidth=2.5,
    label="Actual Capital"
)

plt.plot(
    x_positions,
    prediction_df["RF_Predicted_PaidupCapital"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Random Forest Prediction"
)

plt.plot(
    x_positions,
    prediction_df["NN_Predicted_PaidupCapital"],
    marker="^",
    linestyle="--",
    linewidth=2,
    label="Neural Network Prediction"
)

plt.xticks(
    x_positions,
    prediction_df["Original_Row_Index"].astype(str)
)

plt.xlabel("Original Dataset Row Index")

plt.ylabel("Paid-up Capital")

plt.title(
    "Actual vs Predicted Manufacturing Company Capital",
    fontsize=15,
    fontweight="bold"
)

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

prediction_graph_path = os.path.join(
    PROJECT_DIR,
    "prediction_graph.png"
)

plt.savefig(
    prediction_graph_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("Saved:", prediction_graph_path)

# ============================================================
# STEP 15: FINAL SUMMARY
# ============================================================

print("\n" + "=" * 65)
print("ALL RESULTS GENERATED SUCCESSFULLY")
print("=" * 65)

output_files = [
    "accuracy_graph.png",
    "heatmap.png",
    "comparison_graph.png",
    "results.csv",
    "result_graph.png",
    "prediction.csv",
    "prediction_graph.png"
]

for filename in output_files:

    full_path = os.path.join(
        PROJECT_DIR,
        filename
    )

    if os.path.isfile(full_path):

        size_kb = (
            os.path.getsize(full_path) / 1024
        )

        print(
            f"[SAVED] {filename} "
            f"({size_kb:.2f} KB)"
        )

    else:
        print(f"[MISSING] {filename}")

print("\nPROJECT DIRECTORY:")
print(PROJECT_DIR)

print("\nRandom Forest MAE:", rf_metrics["MAE"])
print("Random Forest RMSE:", rf_metrics["RMSE"])
print("Random Forest R2:", rf_metrics["R2"])

print("\nNeural Network MAE:", nn_metrics["MAE"])
print("Neural Network RMSE:", nn_metrics["RMSE"])
print("Neural Network R2:", nn_metrics["R2"])

print("\nExecution completed.")
```
